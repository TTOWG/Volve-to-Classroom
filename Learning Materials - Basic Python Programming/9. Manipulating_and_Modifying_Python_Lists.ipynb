**List Comprehension**

Notice that the * operator does not do an element-wise multiplication.

If it is intended to multiply every element of a list by 2, then, a *List Comprehension* operation is used.

A _List Comprehension_ statement generates a new list by applying an expression to each element (that meets a condition) of an existing list - it is a short-hand notation of a _for..._ loop that iterates through every element of a list.

If no condition is specified, the _List Comprehension_ applies to all elements in the list.

The code in the diagram above reads thus: for each element in $[1,2,3,4]$; if it is greater than 2, raise the element to power 3; if it is not greater than 2, ignore; ultimately, return a new list of the elements raised to power 3.

```python
[i for i in [1,2,0,3,4] if i!= 0]    # no condition
```

```python
# Doing it with list comprehension

[i**3 for i in [1,2,3,4] if i>2]    # with condition
```

```python
# Doing it without list comprehension

new_list = []
for i in [1,2,3,4]:
    if i > 2:
        new_list.append(i**3)
print(new_list)
```

### List Methods

**General Introduction to Python _Methods_**

Generally, in Python, _methods_ are _functions_ that are specifically tied (built-in) to an object (an instance of a class; such as _class int_, _class str_ or _class list_).

Simply stated, _methods_ are _functions_, only that they are specifically associated with a given instance of an object.

All _methods_ are _functions_, but not all _functions_ are _methods_.

While _functions_ (that are not _methods_) can be called independently and may be applied to any object, _methods_ can only be called on a specific instance of an object.

Think of an independent _fuction_ as an electric power socket that can be used to supply power to just any TV (or even any appliance).

Then, think of a _method_ as a TV remote control that can only be used for a specific TV (not even for just any other TV).

A major difference between independent _funtion_ and _methods_ is in their calling syntax.

Whereas:

+ _functions_ are called as *function_name(first argument, other arguments - if there are)*;
+ _methods_ are called as *object_name.method_name(other arguments - if there are)*.

```python
mantra = 'ttowg'

mantra.upper()

# if upper were to be a function, it will be called thus

# upper(mantra)
```

Another major difference between _functions_ and _methods_ is that:
+ _functions_ effect changes by creating new objects; they do not modify existing objects.
+ _methods_ effect changes by modifying the existing object on which it is called; they typically do not create new objects.

See an example below;

+ my_list is a list of some integers
 + *sorted(my_list)* is a function call that creates a new list (my_list_arranged) where the elements of an my_list are stored after being sorted in ascending order.
 + *my_list.sort()* is a call to _Method sort_ associated with an instance of lists (*my_list*) - it modifies *my_list* by sorting its elements in ascending order without creating a new list object. 

```python
my_list = [5, 2, 8, 1]

# Using the function
my_list_arranged = sorted(my_list)  
print(my_list_arranged)
print(my_list)

# Using the method
my_list.sort()
print(my_list)
```

**Re-visiting the STOIIP computation for discretized reservoirs**

Consider that we also want to generate a list of STOIIP values for all gridblocks in the discretized model, in addition to obtaining the total STOIIP value.

```python
# input statements
Lx = float(input('What is the length of the reservoir in x-direction?\n'))
Ly = float(input('What is the length of the reservoir in y-direction?\n'))
h = float(input('What is the thickness of the reservoir?\n'))
nx = int(input('How many blocks there are in x-axis?\n'))
ny = int(input('How many blocks there are in y-axis?\n'))
boi = float(input('What is the value of initial oil FVF?\n'))

# discretizing the reservoir

delta_x = Lx/nx
delta_y = Ly/ny

# calculating the area per block
area = (delta_x*delta_y)/43560
print(area)

# initializing aggregator variable

total_stoiip = 0
stoiip_list = []

# the 'for' loop
for j in range(1,ny+1):
    for i in range(1,nx+1):
        block_n_order = (nx*(j-1))+i
        print(f'Computing STOIIP for Block{block_n_order}, i.e Block [{i},{j}] now...\n')
        poro = float(input(f'What is the value of porosity for Block {block_n_order}?'))
        sw = float(input(f'What is the value of water saturation for Block {block_n_order}?'))
        block_stoiip = (7758*area*h*poro*(1-sw))/boi
        total_stoiip = total_stoiip + block_stoiip    # aggregation
        stoiip_list.append(block_stoiip)
        
        print(f'\nThe amount of oil in Block {block_n_order}, i.e Block [{i},{j}] is {round(block_stoiip, 2)} STB\n\n')
        print(stoiip_list)
        
    print(f'\nThis is the end of computations for Row {j}\n\n')



# displaying the final results.
print(f'The amount of oil in the entire reservoir is {round(total_stoiip, 2)} STB')
print(stoiip_list)

```

**.extend() method**

Whereas the _.append_ method permits only a single element to be appended to an existing list, the _.extend_ method permits multiple elements (bundled as a list) to be appended to the existing list. This becomes useful when you want to update an existing dataset with some new (multiple) data values

```python
pressures = [4000, 3890, 3503]
more_pressure = [3205, 2956]

print(pressures)
pressures.extend(more_pressure)

print(pressures)

print(more_pressure)
```

Take note that the **_object list_** on which the _.extend_ method is called (_pressures_) gets modified (new elements added to it); but the **_argument list_** passed (as argument) to the call (_more_pressures_) is NOT modified.

**Caution**

Again, note that the _.append_ and _.extend_ methods do not return values (since they don't create new objects); rather, they modify the existing object list in-place. 

This calls for caution: 
+ never execute a statement like: pressures = pressures.extend()

In such a case, the _method_ call returns and assigns **_None_** to the LHS variable; since the LHS variable is the same as the _object list_, the Value _None_ will overwrite the data values stored in the list resulting into a loss of those data values. Clearly, that would be undesirable; except intentional.

```python
pressures = [4000, 3890, 3503]
more_pressure = [3205, 2956]

print(pressures)

pressures = pressures.extend(more_pressure)

print(pressures)
```

While the _List methods_ considered so far are useful for expanding (adding items) to a _list_; sometimes, it is desirable to remove items from _lists_. The next set of _List methods_ to be considered address the need to remove items from _lists_.

**The .pop() method**

An element could be removed from any position in a _list_ using the _.pop()_ method, if the position is known, of course.

Method _.pop()_ **returns the removed value**, and also modifies the object in-place accordingly.

Method _.pop()_ could be useful if you need to use a value in a list and like to remove the used value to avoid re-using it.

The method accept only one argument: the index of the item to be removed (not the value of the item).

**Note**: when the _.pop()_ method is called without specifying an argument, by default, it removes the last item in the list.

**A PE Use-Case**

In reservoir engineering, the initial pressure is recorded at the start of production, and subsequent pressures are measured at different times. Using _.pop()_, we can retrieve and remove the latest pressure reading and compute the pressure drop from the initial pressure.

```python
# Initial pressure is the first element, followed by pressures at later times
pressure_history = [4000, 3800, 3600, 3400, 3200]  # psi

latest_pressure = pressure_history.pop(-1)  # Remove the most recent pressure
initial_pressure = pressure_history[0]  # First recorded pressure

pressure_drop = initial_pressure - latest_pressure  # Compute pressure drop

print(f"Initial Pressure: {initial_pressure} psi")
print(f"Latest Pressure: {latest_pressure} psi")
print(f"Pressure Drop: {pressure_drop} psi")

print(pressure_history)
```

**The _.sort()_ method**

Elements of a list can be sorted either in ascending order of descending order using Method sort.

By default, the method sorts in ascending order; so, users need to pass no argument. However, to sort in descending order, users need to set the keyword argument 'reverse' to True.
