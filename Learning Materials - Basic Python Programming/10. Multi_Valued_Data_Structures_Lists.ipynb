### Lists Objects

A _List_ object is essentially a sequence of values with the following capabilities:
+ It is mutable – the constituent values can be altered/modified in-place (i.e. altered without creating a new _List)
    + Alterations such as changing values, sorting etc.
+ It has variable length – values may be added/deleted to/from the _List_.
+ It can hold any type of values: integers, floats, characters, lists, tuples, dictionaries or mixture of these.

Individual values in a list are called elements or items.

Note that _Lists_ are one-dimensional: i.e. the sequence of values are only in one-dimension (columns); not two dimensions as in rows and columns.

### Indexing Lists to access its elements

Necessarily, there would be need to access individual or group of elements in a list.

Such need to access might be for the purpose of:
+ extraction:
 - of such element(s) for use in an expression.
 - of such element(s) to be assigned to another variable
+ re-assignment: i.e. changing the value of that element

In any case, accessing the element(s) of a list is done by referring to the index (or indices) of the concerned element(s).

**List Indices**

List indices are simply integers that corresponds to the position of the respective elements of lists.

**Note**: 
+ The position starts counting from 0, not 1.
 + The index of first element is 0; 
 + The index of the second element is 1;
 + The index of nth element is n-1 

+ Negative indices counts backward in a list, starting from -1

To access a single element of a list, simply type the name of the list, followed by a pair of square brackets in which the index is specified.

```python
poro = [0.1, 0.25, 0.29, 0.33, 0.23, 0.27, 0.25, 0.1, 0.29, 0.23, 0.33, 0.27]

reser_prop = ['poro', 'perm', 'water_sat', 'oil_sat', 'acreage', 'pay_thick']
```

**Extracting elements to use in an expression**

To extract element(s) of a list to use directly in an expression, simply include the indexing statement in the expression.

New values could be re-assigned to existing list element(s) - the new value replaces the existing value at the index indicated.

```python
print(poro)    # before re-assignment

poro[5] = 0.39     # re-assignment

print(poro)       # after re-assignment
```

### List Slicing - accessing multiple elements

To access multiple elements of a list (a slice of the list), simply type the name of the list, followed by a pair of square brackets in which the **range of indices** is specified.

A range of indices is specified by typing the starting index, followed by a colon, and finally, the index before which the slice must stop.

Generally, a range of indices to access a slice containing $n^{th}$ element through to the $m^{th}$ element is specified as **$n-1:m$**. 

Take note that both $n$ and $m$ here refer to natural element counters (e.g. $3^{rd}$ through to $7^{th}$ in the diagram above)

The index on the left of the colon (i.e. the starting index) is $n-1$ because indices are counted starting from $0$.

The index on the right of the colon is $m$ (not $m – 1$, as expected by reason of the zero start) because Python excludes the last element of any range, so it stops at (m - 1)th position.

```python
poro
```

```python
poro[4:9]
```

```python
poro[1:3] = [0.18, 0.11]       # re-assigning a slice
print(poro)
```

If the left (start) index is not specified, then the slice is taken from the beginning of the list to the indicated stop index.

```python
poro
```

```python
#west_poro = poro[0:4]
west_poro = poro[:4]      # equivalent to poro[0:4] 
west_poro
```

If the right (stop) index is not specified, then the slice is taken from the indicated start index to the end of the list.

```python
east_poro = poro[7:]     # equivalent to poro[4:10]
east_poro
```

If both indices are not specified, then the entire list is returned unsliced.

```python
copy_poro = poro[:]  
copy_poro
```

**Copying Lists**

Indexing an entire list like this is the way to create a copy of the list, with no reference to the original list anymore. That is, changes can be made to the copy without effecting the change in the original list. 

Copying a  list by mere assignment to a new variable still retains reference to the original list; any change in either one (the copy or the original) is refelected in the other.

```python
original = [10, 20, 56]

copy_ref = original        # copying by assigning; retains reference to the original
copy_noref = original[:]      # copying by slicing; no reference to the original 

print(original)
print(copy_ref)
print(copy_noref)

# make a change to original
original[0] = 900

print(original)
print(copy_ref)
print(copy_noref)
```

### List Operations, Methods and Functions

**Concatenation**: 
+ combining multiple lists (side by side) using the + operator.

```python
more_poro = [0.28, 0.19, 0.20, 0.16]

all_poro = poro + more_poro

print(all_poro)
```

**Replication:**
+ replicating all elements of a list a given number of times using the * operator.

```python
perm_left = [120, 114, 150, 170]

all_perm = perm_left*2
print(all_perm)
```

**List Comprehension**

Notice that the * operator does not do an element-wise multiplication.

If it is intended to multiply every element of a list by 2, then, a *List Comprehension* operation is used.

A _List Comprehension_ statement generates a new list by applying an expression to each element (that meets a condition) of an existing list - it is a short-hand notation of a _for..._ loop that iterates through every element of a list.

If no condition is specified, the _List Comprehension_ applies to all elements in the list.

The code in the diagram above reads thus: for each element in $[1,2,3,4]$; if it is greater than 2, raise the element to power 3; if it is not greater than 3, ignore; ultimately, return a new list of the elements raised to power 3.

```python
[i**3 for i in [1,2,3,4]]    # no condition
```

```python
[i**3 for i in [1,2,3,4] if i>2]    # with condition
```

```python
# Doing it without list comprehension

new_list = []
for i in [1,2,3,4]:
    if i > 2:
        new_list.append(i**3)
print(new_list)
```

```python
# List Comprehension
[x*2 for x in all_perm]
```

**.append method**

An extra element (just 1) could be appended to the end of a list using the append method.

This append method is commonly used to append a newly generated output to the overall list of outputs in a for loop.

```python
all_stoiip = [1456, 1738, 1509]    # say these are stoiip values for Blocks 1,2,3

print(all_stoiip)

stoiip_block4 = 1611   # say stoiip for Block 4 just got obtained

all_stoiip.append(stoiip_block4)    # appending stoiip for Block 4 to the nd of the list

print(all_stoiip)

stoiip_block5 = 2045   # say stoiip for Block 5 just got obtained

all_stoiip.append(stoiip_block5)    # appending stoiip for Block 5 to the nd of the list

print(all_stoiip)
```

**.extend method**

An existing list could be extended with multiple elements using the extend method.

```python
pressures = [4000, 3890, 3503]
more_pressure = [3205, 2956]
print(pressures)
pressures.extend(more_pressure)

print(pressures)
print(more_pressure)
```

Please, note that the _.append_ and _.extend_ methods do not return values; rather, they modify the object list in-place (without copying). 

Caution: never do poro = poro.append()

The argument list remains unmodified.

The object list refers to the list on which the method is called while the argument list refers to the list to be appended to (or extended) to the object list.


```python
pressures = [4000, 3890, 3503]
more_pressure = [3205, 2956]
print(pressures)
all_pressures = pressures.extend(more_pressure)

print(pressures)
print(more_pressure)
print(all_pressures)
```

```python
print(pressures)
```

**.insert method**

An element could be inserted into any position in a list (not just the end of the list) using the insert method.

Method insert does not return a value, it only modifies the object in-place.

The method accepts two arguements: the index of the position to be inserted into, and the value to be inserted.

```python
poro = [0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15, 0.42, 0.11, 0.26]
print(poro)


# inserting 0.38 into the '2th' (3rd) position

poro.insert(2, 0.38)
print(poro)
```

**.pop method**

An element could be removed from any position in a list using the pop method, if the position is known, of course.

Method pop returns the removed value, and also modifies the object in-place accordingly.

Method pop could be useful if you need to use a value in a list and like to remove the used value to avoid reusing it – sort of sampling without replacement.

The method accept only one argument: the index of the item to be removed (not the value of the item) 

```python
perm = [120, 150, 400, 387]
print(perm)

# removing the 1th (2nd) item
removed_perm = perm.pop(1)

print(perm)
print(removed_perm)
```

**.remove method**

If the position of the element to be removed is not known, but the value is known; then the remove method can be used to remove it.

However, Method remove only removes the first occurrence of the specified value in the list; subsequent occurrences of that specified value are not removed.

Method remove does not return a value, it only modifies the object in-place.


```python
perm = [120, 150, 400, 387, 600, 150, 300]
print(perm)

# removing the first occurence of 150
removed_perm = perm.remove(150)

print(perm)
print(removed_perm)
```

**del operator**

Both Methods pop and remove accept only one argument; so they can only delete one  element of the list at a time.

To delete multiple elements, use the del operator.

The del operator accepts the indexed list (the list and the range of indices to be deleted as its argument.


```python
poro = [0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15, 0.42, 0.11, 0.26]
print(poro)

del(poro[3:6])

print(poro)
```

**.sort method**

Elements of a list can be sorted either in ascending order of descending order using Method sort.

By default, the method sorts in ascending order; so, users need to pass no argument. However, to sort in descending order, users need to set the keyword argument 'reverse' to True.

The method returns no argument, it modifies in-place


```python
pressures = [4000, 3890, 3503, 3205, 2956]

pressures.sort()

print(pressures)

pressures.sort(reverse = True)

print(pressures)
```

**in operator**

That a value is present (or not present) in a list could be determined using the in operator.

The in operator returns True if the specified value is present in the list, and False, otherwise.

Such in operations might be used in constructing the conditions for if and while statements.

```python
poro = [0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15, 0.42, 0.11, 0.26]

# check if 0.12 is in poro
print(0.12 in poro)

# check if 1000 is in poro
print(1000 in poro)

```

**.index method**

Sometimes, the presence of a value in a list is already ascertained, but there is a need to determine the exact location (index) of the value in the list. Method index could be used for such purpose.

If the specified value occurs more than once, only the index of the first occurrence is returned.

Assignment: how may the indices of all occurrences be obtained?

If the specified value is not present, an error is thrown. So, you may want to preceed the call to .index with a call to the 'in' operator

Method index may be used to implement a find and replace functionality.


```python
poro = [0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15, 0.42, 0.11, 0.26]

print(poro.index(0.37))

to_find = 0.42

if to_find in poro:
    print(f"The value {to_find} is in the {poro.index(to_find)}th position of the list")
else:
    print(f"The value {to_find} is not present in the list")
```
