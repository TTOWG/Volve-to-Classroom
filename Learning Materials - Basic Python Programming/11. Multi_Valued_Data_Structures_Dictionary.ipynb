### The Need for Dictionaries - fast lookup

Imagine that your phone Contacts app stores only your contacts' phone numbers (_values_) without their names (_keys_). That is, your Contacts list is just a _list_ of numbers (maybe arranged sequentially in the order in which you created them).

How frustrasting will it be to look up a particular contact phone number?!

Even if you know the position (index) of that particular contact in the list, you will still need to visually count from the beginning of the list to reach that position.

This is what will happen if you store phone contacts with a Python _list_; recall that _lists_ index the data items with integers.

Would it not be efficient to have a Python data structure that will allow you to **tag** each phone number (data _value_) with the name of the contact (data _key_). 

With this, you only have to scroll through the contact list and be on the look out for the name (_key_) tagged with the desired contact number (_value_). 

Better still, you only have to search (index) the contact list with the name (_key_) to retrieve the desired contact number (_value_).

Enter Python _Dictionaries!!!_

Now, consider a PE case:

You are computing hydrocarbon pore-volumes for gridblocks in a discretized reservoir ($HCPV = BV\phi(1-Sw_i)$).

You have the option of storing the output pore-volumes as a _list_ or as a _dictionary_.

Let's see which option enhances a fast lookup of the output object.

```python
# PE Example:

poro_list = [0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15, 0.42, 0.11, 0.26, 0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15]
swi_list = [0.28, 0.25, 0.17, 0.33, 0.16, 0.23, 0.29, 0.22, 0.32, 0.27, 0.28, 0.25, 0.17, 0.33, 0.16, 0.23, 0.29]
BV = 100

# generating a list of hydrocarbon pore-volumes (BV*porosity*(1-swi))

# Option A - using a list.
    # pre-create the hcpv list before looping
hcpv_list = []
for i in range(len(poro_list)):
    hcpv_val = round(BV*poro_list[i]*(1-swi_list[i]), 2)
    hcpv_list.append(hcpv_val)
print ("============= Option A: using a list=============")    
print(hcpv_list)
print ("\n")
print ("\n")


# Option B - using a dictionary.
    # pre-create the hcpv dictionary before looping
hcpv_dict = {}
ind_list = range(len(poro_list))
for i in ind_list:
    hcpv_val = round(BV*poro_list[i]*(1-swi_list[i]), 2)
    block_id = f'Block {i+1}'
    hcpv_dict[block_id] = hcpv_val

print ("============= Option B: using a dictionary=============")
print(hcpv_dict)
```

### Data Representation in _Dictionaries_: _key-value_ pairs

Simple!

Each data item in _dictionaries_ is a pair of a _value_ and its associated _key_; i.e _key-value_ pair.

The elements in a list are indexed with integers (indices) corresponding to their respective positions in the sequence.

However, the indices (known as _keys_) of a dictionary object are specified alongside the elements (_values_) in the sequence.

That is, the elements (_values_) of a dictionary object are indexed with their respective _keys_ as specified along with the _values_.

This makes the items in a dictionary to be pairs of _key-value_; while the items in a list/tuple are only _values_.

**Keys**

_Keys_ must be of an immutable data type (strings (often) or integers).

_Keys_ have to be unique; i.e no two or more keys should be identical. If two _keys_ are identical, the latest replaces the former.

**Values**

_Values_ can be of any data type.

**PE Example**

A collection of values corresponding to the drainage area (A), payzone thickness (h), porosity (poro), initial water saturation (swi) and oil formation volume factor (boi) for a certain reservoir may be presented as a list or as dictionary thus:

```python
# as a list
resprop_list = [40, 15, 0.3, 0.28, 1.2]

# as a dictionary
resprop_dict = {'area': 40, 'thickness': 15, 'poro': 0.3, 'swi': 0.28, 'boi': 1.2}

print(resprop_list)

print(resprop_dict)
```

### Creating Dictionaries and Assigning _Key-Value_ Pairs

**With Curly Brackets**

A _dictionary_ is simply created by enclosing _key-value_ pairs in **curly brackets**.

A colon (i.e. :) separates a _key_ from its _value_.

And, the _key_value_ pairs are separated with commas.

```python
resprop_dict = {'area': 40, 'thickness': 15, 'poro': 0.3, 'swi': 0.28, 'boi': 1.2}


print(resprop_dict)
```

An empty dictionary, (to be updated later in a 'for' loop) may be defined as {}

**With Function _dict()_**

Function _dict()_ may also be used to define a dictionary (in essence, to convert an object to a dictionary).

Function _dict()_ takes either:
+ keyword arguments (the keywords being the intended _keys_), or 
+ a collection (list or tuple) of pairs; each pair could also be lists or tuples; e.g., a list of tuples.

Also, Function _dict()_ may also be used to create an empty dictionary - with empty argument list, of course.

```python
rock_prop = dict(poro = 0.2, perm = 100)

print(rock_prop)

fluid_prop = dict((['density',0.92], ['fvf', 1.28]))

print(fluid_prop)

stoiip_dict = dict()


print(stoiip_dict)
```

### Accessing Data in Dictionaries

Data _values_ in a dictionary are accessed (both for retrieval and for re-assignment) simply by indexing with the _keys_.

Just like in _lists_ and _tuples_, indexing is done with square brackets.

#### Retrieving data in dictionaries for computation.

This is done by simply featuring the indexed _dictionary_ in expressions.

Racall Function _stoiip_ created in previous modules.

We may re-write the function stoiip such that the five arguments may be passed as a single argument; being a dictionary object containing all the required parameters

In such _dictionary_, the names of the parameters are used as _keys_ for the parameter _values_.

For example: **{'area': 40, 'thickness': 15, 'poro': 0.3, 'swi': 0.28, 'boi': 1.2}**


```python
def stoiip(data):
    N = (7758*data['area']*data['thickness']*data['poro']*(1-data['swi']))/data['boi']
    return round(N, 2)

# calling the function
data_dict = {'area': 40, 'thickness': 15, 'poro': 0.3, 'swi': 0.28, 'boi': 1.2}
stoiip(data_dict)
```

#### Re-assigning values in dictionaries

It may be neccessary to re-assign values to existing _keys_ in a _dictionary_.

Imagine monitoring oil pipelines, and after maintenance on one of the pipelines, its flow rate increases; so, you have to update it.

```python
# Flowrates before maintenance
pipeline_flow = {
    "Line A": 2500,
    "Line B": 3200,
    "Line C": 2800
}

print(pipeline_flow)

# Updating flowrate for Line B after maintenance
pipeline_flow["Line B"] = 3500

print(pipeline_flow)
```

#### Indexing _dictionaries_ with non-existent _keys_ - to assign new _values_

Unlike _lists/tuples_, _dictionaries_ permit out-of-range indexing (i.e. indexing with _keys_ not yet existing in the _dictionary_. 

Of course, such indexing is not for retrieval purposes, but for assignment purposes only.

In such case, a new data item (a _key-value_ pair corresponding to the specified _key_ and the assigned _value_) is created.

It is with such out-of-range indexing that a _dictionary_ can be updated in a loop; _dictionaries_ do not have the insert and append methods.

```python
rock_prop = {'poro': 0.2, 'perm': 100}

print(rock_prop)

rock_prop['swi'] = 0.32

print(rock_prop)

rock_prop['soi'] = 0.78

print(rock_prop)
```

```python
# PE Example:

poro_list = [0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15, 0.42, 0.11, 0.26, 0.49, 0.45, 0.26, 0.12, 0.37, 0.32, 0.15]
swi_list = [0.28, 0.25, 0.17, 0.33, 0.16, 0.23, 0.29, 0.22, 0.32, 0.27, 0.28, 0.25, 0.17, 0.33, 0.16, 0.23, 0.29]
BV = 100

# generating a list of hydrocarbon pore-volumes (BV*porosity*(1-swi))

# pre-create the hcpv dictionary before looping
hcpv_dict = {}     # empty dictionary

for i in range(len(poro_list)):
    hcpv_val = round(BV*poro_list[i]*(1-swi_list[i]), 2)
    block_key = f'Block {i+1}'
    hcpv_dict[block_key] = hcpv_val


print(hcpv_dict)
```

### Looping through Dictionaries

Just like _lists_ and _tuples_; the items of a _dictionary_ can be looped through.

In such loops, the _dictionary keys_ are the default iterator variables (not the _values_)

If the dictionary _values_ are needed in the loop, they are simply accessed by indexing with the _keys_ (which are the respective iterator variable).

In essence, a loop through a _dictionary_ is a loop through its _keys_; except otherwise constructed.


```python
rock_prop = {'poro': 0.2, 'perm': 100}

for i in rock_prop:
    
    print(i)
```

```python
rock_prop = {'poro': 0.2, 'perm': 100}

for i in rock_prop:
    
    print(rock_prop[i])
```
