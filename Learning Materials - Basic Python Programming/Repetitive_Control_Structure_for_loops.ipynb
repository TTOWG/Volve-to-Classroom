A *for...* loop is capable of repeating/iterating a code block as many times as there are elements in the collection being looped over. 



### Iterable Objects

An iterable is a sequence object that is capable of presenting its members one at a time when it is called upon.

In essence, it is a collection of 'things'; like sets, in mathematics.

Examples:
+ Lists
+ Tuples
+ Dictionaries

```python
# Examples of iterables:

# a list
perm = [100, 45, 98]

# a tuple
components = ('methane', 'ethane', 'propane', 'butane', 'pentane', 'hexane', 'heptane+')
```

### Iterator Variable

When looping (iterating) through an iterable object, there is a need to keep track of the current position; a need to know whose turn is it among the elements of the iterable object.

The **Iterator Variable** is the variable created to perform this tracking duty. It keeps track by asigning the value of the current element to itself.

In the first iteration, it takes on the value of the first element; in the second iteration, it drops the first value of the first element and takes on the value of the second element; so on, so forth.

So, the iterator variable is a dynamic variable - with its value changing as iterations continue.

```python
comp_iterator = iter(components)
```

```python
next(comp_iterator)
```

### Structure of a 'for...' loop

A 'for...' loop basically has a header and a body.

**The Header**
+ Starts with the keyword: 'for'
+ Next, the iterator variable
+ Next, another keyword: 'in'
+ Then, the iterable object to be looped through
+ Ends with a colon ':'

**The body**
+ The code line(s) to be repeated.
+ Every line has to be indented under the header

```python
# Example of a 'for...' loop

for comp in components:
    print(comp)
```

### Counting with 'for...' loops

One of the most basic use of 'for...' loops is as counters.

Looping through an iterable, the number of elements in the iterable can be counted.

This is possible by setting (initializing) a counter variable to zero just before the loop starts, and then incrementing the variable by 1 during each iteration.

```python
components = ('methane', 'ethane', 'propane', 'butane', 'pentane', 'hexane', 'heptane+')

# initialize counter variable to zero;
# a counter is naturally equal to zero just before the counting begins
n_comps = 0

# the 'for' loop
for comp in components:
    print(comp)
    n_comps = n_comps + 1  #incrementing
    print(n_comps)
    

print(f'There are {n_comps} components in the fluid mixture')     
```

### Aggregating with 'for...' loops

Sometimes, it is desirable to do more than just counting. You may want to cumulatively sum up the elements through the iterations.

```python
well_rates = [150, 130, 90, 140, 100, 120, 110]

#initialize the aggregator variable
total_rate = 0
counter = 0

for rate in well_rates:
    total_rate = total_rate + rate
    counter = counter + 1
    print(f'So far, the cumulative rate is {total_rate} STB/D')

print(f'Total rate is {total_rate} STB/D')

average_rate = total_rate/counter

print(f'Average rate is {average_rate} STB/D')
```

### Variations of 'for...' loops

The iterable to be looped through may not necessarily be created before the 'for' loop header

It may be directly referenced right in the header.

```python

#blocks = [1,2,3,4,5]

#for block in blocks:
#    print(f'This is Block {block}')

#The code chunk above is equivalent to this below:

for block in [1,2,3,4,5,6]:
    print(f'This is Block {block}')
```

Also, if the iterable to be looped through is a logical sequence of integers, it could be created using Function 'range'

The function generates  a list of consecutive (or evenly-spaced) integers, starting (by default) from zero and ending 1 short of its argument. For this reason, always add 1 to the intended stoping value.

```python
list(range(6))
```

If it is desirable to start at any other number other than 0, then the starting point is specified with the first argument of function.

```python
list(range(1, 6))
```

Ultimately, Function range could be used to generate an equally spaced (arithmetic progression) sequence of integers.

```python
list(range(1, 30, 5))  #Nth = a+(n-1)d
```

Such calls to Function 'range' may be made directly at the header of a 'for...' loop.

```python
#for block in [1,2,3,4,5]

for block in range(1,6):
    print(f'This is Block {block}')
```
