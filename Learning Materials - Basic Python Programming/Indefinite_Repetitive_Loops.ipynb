### Definite Loops versus Indefinite Loops

Whereas, a definite (_for..._) loop repeats a block of statement(s) for each item in a **collection** of such items passed to it; no such collection of items is needed for an indefinite (_while..._) loop. Rather, an indefinte loop repeats a block of statement(s) as long a **condition** (logical expression) passed to it is true. The repetition stops when the condition becomes false. The fact that the number of iterations a '_while..._' loop will perform is not known prior to implementation is the reason it is called indefinite loop.

The while... loops finds applications in cases where a task need to be performed for a series of dynamically-changing values of a variable (like pressure) until that variable falls below/above a specified cut-off value.

Whereas, the execution of a definite loop goes thus:

   + Check if there are still elements in the given collection
   + If there are still elements, execute the block of statement(s) and return to check the collection again.
   + If there are no more elements, exit the loop and continue with statements(s) outside (below) the loop

The execution of an indefinite loop goes thus:
    
   + Evaluate the specified condition
   + If the condition evaluates to True, execute the block of statement(s) and return to evaluation step.
   + If condition evaluates to False, exit the loop and continue with statements(s) outside (below) the loop.

### Elements of a _while..._ Looop

The basic elements of the _while_ loop is thus:

* The header:
 + starts with the _while_ keyword,
 + then, the condition for continuation 
     - when true, loop continues;
     - when false, loop terminates
 + ends with a colon character

* The body:
 + the code block that should be repeated as long as the condition remains true.
 


```python
# initialize the iterator variable:

i = 1

# the 'while' loop:
while i < 11:
    print(f'Bring in Student {i}')
    i = i+1 # incrementing the iterator variable
    

print('No more space in the class')
```

### The _while..._ Loop _Condition_ and the _Iterator_ Variable

The ‘while’ loop condition is constructed as Boolean expression.

Typically, the condition involves an iterator variable that is:
  
+ Initialized outside (before) the loop to a value that would make the condition evaluates to True (to start with)
   
    - if not initialize, it cannot feature in the condition
       
    - if initialize to a value that makes the condition to evaluate to False, then, the loop will not start at all.
        
+ Updated in the loop (during each iteration), i.e., its value get changed by some simple arithmetic operations such as addition and subtraction
     
    - such changes would eventually get the iterator variable to a value that would make the condition to evaluates to False and the loop gets terminated.
        
    - if not updated, it will remain at the initial value, and the condition would always evaluates to True, and so the loop will run 'forever' - that is known as **infinite** loop.


```python
initial_pressure = 4000

current_pressure = initial_pressure
bubble_pressure = 2800

step = 200     # decremental step (mimicks pressure drops in reservoirs)
```

```python
while current_pressure >= bubble_pressure:
    
    print(f'At Pressure {current_pressure} psi, the reservoir is still undersaturated')
    
    print('Computing...')
    
    current_pressure = current_pressure - step    # decrementing current pressure

# At the end of loop:
print(f'Undersaturated computations ends at {current_pressure + step}')
print('Proceeding to saturated regime')
```

### A PE application of _while..._ loop: updating oil formation volume factor

In scripting an undersaturated oil reservoir simulator, computations are to be performed for series of decreasing pressure values until pressure reduces to a value below bubble-point pressure.

**Updating oil formation volume factor ($B_o$), for undersaturated reservoir simulation**

The analytical equation for computing $B_o$, at pressures above bubble point is given thus:

$$𝐵_𝑜= 𝐵_{ob} 𝑒^{[c_o (P_b−P)]}$$

There is a need to write statement that would not just compute $B_o$ at a single pressure value, but at series of decreasing pressure values until pressure decreases to bubble point pressure, at which point, the simulation must be terminated.

```python
# input statements
p = float(input('What is the value of reservoir initial pressure?\n'))
pb = float(input('What is the value of reservoir bubble-point pressure?\n'))
bob = float(input('What is the value of oil FVF at bubble-point pressure?\n'))
co = float(input('What is the value of oil compressibility?\n'))
pressure_step = float(input('By how much should pressure be decremented?\n'))
```

```python
# importing needed library
import math
```

```python
# The 'while' loop
while p >= pb:
    bo = bob*(math.exp(co*(pb-p)))
    print(f'The value of oil FVF at {p} psi is {round(bo, 4)} RB/STB')
    p = p - pressure_step  
```

```python
# continuing after the 'while' loop
print('Bubble-point pressure reached! End of simulation')
```
